```python
numbers = [2, 4, 6, 8, 10, 12]

even = [x for x in numbers if x % 2 == 0]

squares = [x**2 for x in even]

result = [x for x in squares if x > 100]

print("Even numbers:", even)
print("Squares:", squares)
print("Squares greater than 100:", result)
```

```python
words = ["cat", "apple", "dog", "banana", "sun"]

result = [x.upper() for x in words if len(x) > 3]

print("Result:", result)
```

```python
numbers = [2, 3, 4, 5, 6, 7, 9, 10]

result = [x**2 for x in numbers if x % 2 == 0 or x % 3 == 0]

print("Result:", result)
```

```python
numbers = [10, 15, 20, 25, 30]

added = [x + 10 for x in numbers]

result = [x for x in added if x > 30]

print("After adding 10:", added)
print("Elements greater than 30:", result)
```

```python
numbers = [-5, 10, -15, 20, -25, 30]

result = [abs(x) for x in numbers if x < 0]

print("Absolute values:", result)
```

```python
words = ["pen", "apple", "sun", "book", "moon", "fan"]

result = [len(x) for x in words if x.endswith('n')]

print("Lengths of words ending with n:", result)
```

```python
numbers = [1, 4, 8, 9, 12, 16, 20, 25, 30, 36]

result = [x for x in numbers if int(x**0.5)**2 == x]

print("Perfect squares:", result)
```
