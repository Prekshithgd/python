```python
# Employee Salary Processing

employees = [
    (101, "Rahul", 30000, "HR"),
    (102, "Ram", 40000, "IT"),
    (103, "Amit", 25000, "Sales"),
    (104, "Raju", 50000, "Finance")
]

salary_details = {}

for emp in employees:
    emp_id, name, basic, dept = emp

    hra = basic * 0.20
    da = basic * 0.10
    pf = basic * 0.12

    net_salary = basic + hra + da - pf

    salary_details[emp_id] = {
        "Name": name,
        "Department": dept,
        "Basic Salary": basic,
        "HRA": hra,
        "DA": da,
        "PF": pf,
        "Net Salary": net_salary
    }

# Display employee details
for emp_id, details in salary_details.items():
    print("\nEmployee ID:", emp_id)
    for key, value in details.items():
        print(key, ":", value)

# Find highest-paid employee
highest = max(
    salary_details.items(),
    key=lambda x: x[1]["Net Salary"]
)

# Calculate average net salary
total = sum(
    emp["Net Salary"] for emp in salary_details.values()
)

average = total / len(salary_details)

print("\nHighest Paid Employee:")
print("Name:", highest[1]["Name"])
print("Net Salary:", highest[1]["Net Salary"])

print("\nAverage Net Salary:", average)
```
