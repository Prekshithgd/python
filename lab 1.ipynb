```python
number = 5
factorial = 1

for i in range(1, number + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

```python
number = 29
is_prime = True

if number <= 1:
    is_prime = False
else:
    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

if is_prime:
    print("Prime")
else:
    print("Not Prime")
```
